# Pydantic

```python
from pydantic import BaseModel, EmailStr
```

```python
class User(BaseModel):
    name: str
    email: EmailStr
    acc_id: int
```

```python
user = User(
    name="tarun",
    email="tag@mail.com",
    acc_id=42
)
```

#### json_string / JSON Serialization

```python
user_json_str = user.model_dump_json() # user.json() deprecated

print(user_json_str, "\ntype: ", type(user_json_str))
```

#### dictionary

```python
user.model_dump()  # user.dict() deprecated
```

#### json to pydantic model

```python
user2_str_json = '{"name":"tarun","email":"tag@mail.com","acc_id":42} '

print("type: ", type(user2_str_json))

user_2 = User.model_validate_json(user2_str_json) # User.parse_raw() deprecated
print(user_2, "\n\npydantictype of user_2: ", type(user_2))
```

#### Validation

```python
from pydantic import validator, field_validator, ValidationError
```

```python
user_invalid_id = User(
    name="arjun",
    email="k@joe.io",
    acc_id="invalid_string"
)
```

```python
# custom validator
class User(BaseModel):
    name: str
    email: EmailStr
    acc_id: int
    @field_validator("acc_id") # @validator deprecated
    @classmethod
    def validate_acc_id(cls, value: int) -> int:
        if value <= 0:
            raise ValueError(f"acc_id must be positive integer: {value}.")
        return value

user_valid_id = User(
    name="karn",
    email="dosti@imp.ip",
    acc_id=24
)

print(f"Successful! for user: {user_valid_id.name} with id: {user_valid_id.acc_id}.\n")

try:
    user_invalid_id_custom = User(
        name="duryodhan", 
        email="naked@nope.ip", 
        acc_id=-12)
except ValidationError as e:
    print("Error: ", e)

print(f"Successfull! for user: {user_invalid_id_custom.name} with id: {user_invalid_id_custom.acc_id}")
```
